```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
G = nx.Graph()
```

```python
G.add_nodes_from([1,2,3,4,5])
```

```python
G.add_edges_from([(1,2),(1,4),(1,5),(2,4),(2,6),(3,4),(5,6)])
```

```python
nx.draw(G, with_labels=True, node_color="#F4ABAA",node_size=600)
```
